### Importación de paquetes

```python
# Paquetes de Interfaz gráfica
import tkinter as tk
from tkinter import ttk
from PIL import Image
import customtkinter
```

### Base de Conocimiento

```python
# Enfermedades
e1 = "Gripe"
e2 = "Resfriado"
e3 = "Neumonía"

# Síntomas
s1 = "Fiebre"
s2 = "Tos"
s3 = "Congestión nasal"
s4 = "Ardor de Garganta"
s5 = "Dificultad respiratoria"

# Probabilidades marginales
p1 = 0.2 # probabilidad de gripe
p2 = 0.7 # probabilidad de resfriado
p3 = 0.1 # probabilidad de neumonía

# Probabilidades condicionales
p11 = 0.8 # Probabilidad de fiebre dado gripe
p21 = 0.9 # Probabilidad de tos dado gripe
p31 = 0.6 # Probabilidad de congestión nasal dado gripe
p41 = 0.5 # Probabilidad de ardor de garganta dado gripe
p51 = 0.3 # Probabilidad de dificultad respiratoria dado gripe
p12 = 0.3 # Probabilidad de fiebre dado resfriado
p22 = 0.7 # Probabilidad de tos dado resfriado
p32 = 0.9 # Probabilidad de congestión nasal dado resfriado
p42 = 0.8 # Probabilidad de ardor de garganta dado resfriado
p52 = 0.1 # Probabilidad de dificultad respiratoria dado resfriado
p13 = 0.3 # Probabilidad de fiebre dado neumonía
p23 = 0.7 # Probabilidad de tos dado neumonía
p33 = 0.9 # Probabilidad de congestión nasal dado neumonía
p43 = 0.8 # Probabilidad de ardor de garganta dado neumonía
p53 = 0.1 # Probabilidad de dificultad respiratoria dado neumonía
```

### Motor de Inferencia

```python
def diagnosticar (sintomas):
    # Inicialmente si el subconjunto de síntomas es vacío, son las probabilidades a priori
    p_e1_sintomas = p1
    p_e2_sintomas = p2
    p_e3_sintomas = p3

    # Probabilidad conjunta de síntomas identificados dada cada enfermedad
    p_sintomas_e1 = 1
    p_sintomas_e2 = 1
    p_sintomas_e3 = 1

    # Si el síntoma 'fiebre' ha sido identificado, se actualizan las probabilidad conjuntas
    if s1 in sintomas:
        p_sintomas_e1 *= p11
        p_sintomas_e2 *= p12
        p_sintomas_e3 *= p13

    # Si el síntoma 'tos' ha sido identificado, se actualizan las probabilidad conjuntas
    if s2 in sintomas:
        p_sintomas_e1 *= p21
        p_sintomas_e2 *= p22
        p_sintomas_e3 *= p23

    # Si el síntoma 'congestión nasal' ha sido identificado, se actualizan las probabilidad conjuntas
    if s3 in sintomas:
        p_sintomas_e1 *= p31
        p_sintomas_e2 *= p32
        p_sintomas_e3 *= p33

    # Si el síntoma 'ardor de garganta' ha sido identificado, se actualizan las probabilidad conjuntas
    if s4 in sintomas:
        p_sintomas_e1 *= p41
        p_sintomas_e2 *= p42
        p_sintomas_e3 *= p43

    # Si el síntoma 'dificultad respiratoria' ha sido identificado, se actualizan las probabilidad conjuntas
    if s5 in sintomas:
        p_sintomas_e1 *= p51
        p_sintomas_e2 *= p52
        p_sintomas_e3 *= p53

    # Cálculo del denominador en el Teorema de Bayes
    p_sintomas = p_sintomas_e1 * p1 + p_sintomas_e2 * p2 + p_sintomas_e3 * p3

    # Cálculo de las probabilidades a posteriori (Teorema de Bayes)
    p_e1_sintomas = (p_sintomas_e1 * p1) / p_sintomas
    p_e2_sintomas = (p_sintomas_e2 * p2) / p_sintomas
    p_e3_sintomas = (p_sintomas_e3 * p3) / p_sintomas

    # Devuelve una lista con las probabilidades calculadas
    return [p_e1_sintomas, p_e2_sintomas, p_e3_sintomas]
```
